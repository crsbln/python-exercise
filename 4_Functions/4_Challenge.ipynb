For the first part of the assignment, we're going to get some practice taking something we've already written and translating it to a function. 

In continuation of prior assignments, 

1. Write a function that computes the factorial for an inputted number.  

2. Write a function that determines whether or not an inputted number is prime, and then prints 'The number you inputted is a prime/ not a prime number.' depending on what your script finds (note that this means putting a `print` in front of the function when testing it will be redundant for this case). 

```python
# Part 1.1: factorial, e.g. 5! = 5 * 4 * 3 * 2 * 1 = 120.
def factorial(n):
    if n < 0:
        return None  # not defined for negative numbers
    result = 1
    for number in range(2, n + 1):
        result *= number
    return result  # 0! and 1! stay 1
```

```python
# Five tests, including the edge cases 0, 1 and a negative number.
print(factorial(0))
print(factorial(1))
print(factorial(5))
print(factorial(10))
print(factorial(-3))
```

```python
# Part 1.2: a prime can only be divided by 1 and itself.
def is_prime(n):
    if n < 2:
        return False  # 0, 1 and negatives are not prime
    # Testing divisors up to the square root of n is enough.
    divisor = 2
    while divisor * divisor <= n:
        if n % divisor == 0:
            return False  # found a divisor: not prime
        divisor += 1
    return True


def prime_message(n):
    if is_prime(n):
        print("The number you inputted is a prime number.")
    else:
        print("The number you inputted is not a prime number.")
```

```python
# Five tests: 1 (edge case), 2 (smallest prime), 9, 17, 97.
for number in [1, 2, 9, 17, 97]:
    print(number, end=": ")
    prime_message(number)
```

```python
# Part 2.1: split() with no argument splits at any run of spaces.
def count_words(text):
    return len(text.split())


print(count_words("This is a test string"))
print(count_words("Hello"))
print(count_words(""))  # edge case: no words
print(count_words("  extra   spaces  here "))
print(count_words("one two three four five six"))
```

```python
# Part 2.2: same, but the caller picks the delimiter.
# Empty pieces (from two delimiters in a row) do not count as words.
def count_words_by(text, delimiter=" "):
    pieces = text.split(delimiter)
    return len([piece for piece in pieces if piece != ""])


print(count_words_by("This is a test string"))
print(count_words_by("red,green,blue", ","))
print(count_words_by("a-b--c", "-"))
print(count_words_by("", ","))
print(count_words_by("no delimiter here", ";"))
```

```python
# Part 2.3: the length of each word.
def word_lengths(text, delimiter=" "):
    return [len(word) for word in text.split(delimiter)]


print(word_lengths("This is a test string"))  # [4, 2, 1, 4, 6]
print(word_lengths("this. has punctuation!"))
print(word_lengths("red,green,blue", ","))
print(word_lengths("single"))
print(word_lengths("a b c"))
```

```python
# Part 2.4: re-use is_prime() from Part 1 for every number up to n.
def primes_up_to(n):
    return [number for number in range(2, n + 1) if is_prime(number)]


print(primes_up_to(1))
print(primes_up_to(2))
print(primes_up_to(10))
print(primes_up_to(30))
print(len(primes_up_to(100)), "primes up to 100")
```

```python
# Part 2.5: "yes" or "no" for each number.
def divisible_yes_no(numbers, divisor):
    answers = []
    for number in numbers:
        if number % divisor == 0:
            answers.append("yes")
        else:
            answers.append("no")
    return answers


print(divisible_yes_no([10, 25, 36, 12, 20], 5))
print(divisible_yes_no([1, 2, 3, 4], 2))
print(divisible_yes_no([], 3))
print(divisible_yes_no([0, 7, 14], 7))
print(divisible_yes_no([9, 10], 1))
```

```python
# Part 2.6: keep the words whose last letter matches.
def ends_with_letter(words, letter):
    return [word for word in words if word.endswith(letter)]


words = ["I", "am", "in", "love", "with", "Python"]
print(ends_with_letter(words, "n"))  # ['in', 'Python']
print(ends_with_letter(words, "e"))
print(ends_with_letter(words, "z"))
print(ends_with_letter(words, "I"))
print(ends_with_letter([], "a"))
```

```python
# Part 2.7: the positions of the words that contain the substring.
def indices_containing(words, substring):
    return [i for i, word in enumerate(words) if substring in word]


print(indices_containing(["This", "is", "an", "example"], "is"))
print(indices_containing(["This", "is", "an", "example"], "a"))
print(indices_containing(["cat", "dog"], "x"))
print(indices_containing(["Is", "is"], "is"))  # case matters
print(indices_containing([], "is"))
```

1. Let's build a calculator for figuring out how much you owe in taxes (and by calculator, we mean a function). 

Write a function that takes in a list of tuples, where each tuple contains two values as well as an income to compute the taxes on (so your function should accept two arguments). 

For the tuple list, the first value of a tuple will be an income upper bound, and the second will be a tax rate for all income up to the given income bound. 

You need to build a calculator that will calculate the tax for all income up to each income bound (if it goes up that high) for the given tax rate. You can assume that the list of tuples will be sorted by the income bound (e.g. the first value in the tuple), such that the lowest income bound will be first, and the highest last (see below for an example).  

- As an example, let's say my tax info is `[(50000, 0.08), (100000, 0.10), (150000, 0.15)]`. This means that the first 50k of income is taxed at 8% (we wish!), the second 50k at 10% (yeah right!), and the rest at 15% (note here: that any income which comes after the 150k is taxed at 15%. Therefore, giving a highest upper bound is misleading because everything above 100k will be taxed at 15%). 
 
- So, if the inputted income was 170k, then my taxes would be:
    - 50 * 0.08 + 50 * 0.10 + 70 * 0.15 = 19.5k. 
 
- And if the inputted income was 70k, then my taxes would be:
    - 50 * 0.08 + 20 * 0.10 = 6k. 
 
 You should write your function to be generalized to accept any kind of list of tuples and any inputted income (so it'll accept two arguments).  

2. Now modify your solution to accept a list of tuples that is not sorted by the income bound. 

- **Hint**: Trying using the `sorted` function and working with the `key` argument (a [lambda](https://www.educative.io/blog/python-lambda-functions-tutorial) function will probably be helpful here). 

```python
# Extra 1: tax per bracket. brackets = [(upper_bound, rate), ...]
def calculate_tax(brackets, income):
    tax = 0
    lower = 0
    for upper, rate in brackets:
        if income > lower:
            # Only the part of the income inside this bracket.
            taxable = min(income, upper) - lower
            tax += taxable * rate
        lower = upper
    # Income above the highest bound keeps the last rate.
    if income > lower:
        tax += (income - lower) * brackets[-1][1]
    return round(tax, 2)


brackets = [(50000, 0.08), (100000, 0.10), (150000, 0.15)]
print(calculate_tax(brackets, 170000))  # 19500.0, as in the task
print(calculate_tax(brackets, 70000))  # 6000.0, as in the task
print(calculate_tax(brackets, 0))
print(calculate_tax(brackets, 50000))
print(calculate_tax(brackets, 150000))
```

```python
# Extra 2: sort the brackets first. key= tells sorted() to compare
# only the first value (the bound) of each tuple.
def calculate_tax_unsorted(brackets, income):
    ordered = sorted(brackets, key=lambda pair: pair[0])
    return calculate_tax(ordered, income)


shuffled = [(150000, 0.15), (50000, 0.08), (100000, 0.10)]
print(calculate_tax_unsorted(shuffled, 170000))
print(calculate_tax_unsorted(shuffled, 70000))
print(calculate_tax_unsorted(shuffled, 120000))
print(calculate_tax_unsorted([(10, 0.5)], 20))
print(calculate_tax_unsorted(shuffled, 0))
```
