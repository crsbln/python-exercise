Up to this point, we have just worked with functions that return the same output every time. But what if we want a function to act in different ways depending on some input? 

As we've hinted at, functions can be defined so that their behavior changes depending on what values are **passed** to them. You can pass any data structure to a function so long as it expects the right number of them. Values that are passed to a function are called **Arguments**.

How does a function "expect" arguments to be passed to it? 

In the parentheses of a functions definition, we put the names of variables that we expect a user (of our function) to pass to it. We call these special variables **Parameters**. This is where functions get their flexibility. When you define a function with a certain number of `parameters`, you can then *refer* to those parameters within the body of the function. 

The parameters are set when a user passes arguments to the function, so they are actually controlling what happens inside the function!

Let's look at how we can make a function flexible by changing the `get_evens()` function that we defined in the first notebook. 

Instead of creating a list of evens from 0 to 10 every time, let's have `get_evens()` build an arbitrarily sized list of evens, from 0 to a number the user passes in. Using our current `get_evens()` as a base, we'll begin by adding a parameter `n` to the function definition.

**Note**: This parameter will control the size of the evens list that our function builds. 

````python
def get_evens(n):
    evens = []
    for element in range(n):
        if element % 2 == 0:
            evens.append(element)
    return evens
````

You can declare what data types you arguments must have by using the following syntax:

```python
def my_function(arg: int) # Would require an integer to be passed as the argument or an error will occur
```
This works for any data type and serves as a means to control what goes into the function. 

For our `get_evens` function, the same can be applied.

````python
def get_evens(n: int):  # noqa: F811
    evens = []
    for element in range(n):
        if element % 2 == 0:
            evens.append(element)
    return evens
````

With this implementation of our function we can now pass in an arbitrary number to our function call, and then we will search for evens in a `range()` built with that arbitrary number. 

How exactly does this work? Well, by adding explicitely one parameter we've told Python that our function should expect one and only one argument. Get it. 

When we call the function and pass in that argument, it will get assigned to whatever name we have given in the function definition - `n` in this case. Then, anytime we reference that parameter, `n`, within the function, it will be the value that was passed to the function. 

Let's check out a couple of different calls to this function and see what they return.

We can also build it so that our parameter gets a **default value** if the function is called without an argument passed in. 

This is useful if we want to build our function to have some default behavior but still allow users to pass in arguments that change the default behavior, or build off of it somehow. 

How do we specify a default parameter value for a function? 

It's actually pretty simple. In the function definition itself, we just place an equals sign (`=`) after the parameter name:

```python
def my_function(arg=42) # Sets the default value to the integer 42
def my_function(arg: str="My Parameter") # This example we have used a string, but remember! any value, any data-type!
```

As you could of guessed, on the right side of the equals sign is the default value that we want to specify.

**Note**: Python formatting convention dictates that there should be no spaces surrounding equals signs used in this way.

````python
def get_evens(n=15):
    evens = []
    for element in range(n):
        if element % 2 == 0:
            evens.append(element)
    return evens
````

````python
get_evens()
````

````python
get_evens(15)
````

````python
get_evens(10)
````

````python
get_evens(20)
````

Here, we've specified the default value for `n` to be 15. That is, if no value is passed for `n` it gets assigned the value of 15 by default. 

You'll notice in the first function call to `get_evens()` where we pass no arguments, we get the same output as if we pass in the value 15. This makes sense since we set 15 as the default. 

Meanwhile, when we pass in other values, we get the same results as we saw when no default value was set. This is the point of setting a default parameter value - if the caller of the function specifies a value for that argument, then that is the value used in the function; otherwise, the specified default parameter value is used.

We can also define a function with multiple parameters, and then pass in multiple arguments for those parameters when calling the function. 

Similar to specifying a default value for a single parameter, we can specify default values for multiple parameters. 

Let's first modify our function so that we by default return a list of evens from the user inputted range, defined by `n`, but also give the user the option to input a different divisor (instead of 2 for the evens) that will then return numbers in the inputted range that are divisible by the inputted divisor (i.e. the multiples of that number). 

We'll also change our function's name and the name of the returned list (`evens`) so that they become more descriptive as our function is no longer outputting just evens.  

````python
def get_multiples(n=15, divisor=2):
    multiples_lst = []
    for element in range(n):
        if element % divisor == 0:
            multiples_lst.append(element)
    return multiples_lst
````

````python
get_multiples()
````

````python
get_multiples(15)
````

````python
get_multiples(5, 2)
````

````python
get_multiples(15, 4)
````

````python
get_multiples(50, 3)
````

````python
get_multiples(100, 9)
````

As you can see from the first three examples above, the output of our function looks the same as `get_evens()` - by default we still output a list of the even numbers up to 15, and when we pass in 15 as the value of `n` and 2 as the value of `divisor`, we also output a list of the evens up to 15. 

The other function calls give us access to the new, `generalized`, version of `get_evens()`, just as we wanted.

**Important**: Let's take a quick look at a syntactic "rule" that we have to follow when we define functions with default values. 

When we do this, we have to make sure that any parameters we are giving default values are **after** any parameters that we are not giving default values. 

Look at the two examples below...

````python
def get_multiples(n, divisor=2):
    multiples_lst = []
    for element in range(n):
        if element % divisor == 0:
            multiples_lst.append(element)
    return multiples_lst
````

````python
# Intentional error example kept for the exercise.
# Uncomment the lines below to see the error.
# It should raise SyntaxError.
#
# def get_multiples(n=5, divisor):
#     multiples_lst = []
#     for element in range(n):
#         if element % divisor == 0:
#             multiples_lst.append(element)
#     return multiples_lst
````

The above code blocks demonstrate this "rule". 

In the first case, we defined our parameter that has a default value (which is only one, `divisor`) after defining our parameter that doesn't have default values (which is only one, `n`). In this case, everything worked fine! 

In the second case, we defined a parameter with a default value before a parameter without a default value. That's a NO NO! And Python lets us know know.

**Parameters vs Arguments**

As a quick reminder, it may seem like the terms "parameter" and "argument" are being used seemingly interchangeably. 

These two terms have specific and distinct definitions:

- A **parameter** is the name of a variable given in a function definition
- An **argument** is the value that is passed to a function when it is called

````python
# Valid: 2, 3, 5, 6. Invalid: 1 and 4.
# Rule: a parameter WITH a default may not come before one without.
definitions = {
    "my_func1": "def my_func1(var1='Hello', var2):\n    pass",
    "my_func2": "def my_func2(var1, var2='Hello'):\n    pass",
    "my_func3": "def my_func3(var1, var2=35):\n    pass",
    "my_func4": "def my_func4(var1=35, var2):\n    pass",
    "my_func5": "def my_func5(var1=35, var2='Hello'):\n    pass",
    "my_func6": "def my_func6(var1, var2):\n    pass",
}

# compile() checks the code without running it, so a SyntaxError
# can be caught and printed instead of stopping the notebook.
for name, source in definitions.items():
    try:
        compile(source, name, "exec")
        print(f"{name}: valid")
    except SyntaxError as error:
        print(f"{name}: invalid ({error.msg})")
````
